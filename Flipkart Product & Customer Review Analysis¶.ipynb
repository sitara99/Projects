```python
# Product with the highest ratings
import seaborn as sns
import matplotlib.pyplot as plt

# Find product(s) with highest rating
highest_rated = df[df['Rate'] == df['Rate'].max()]

# Count occurrences of each product with highest rating
top_products = highest_rated['Product_name'].value_counts().head(10)

# Plot
sns.barplot(x=top_products.values, y=top_products.index, palette="viridis")
plt.title("Top Products with Highest Rating")
plt.xlabel("Count of Reviews")
plt.ylabel("Product Name")
plt.show()
```

```python
# Price VS Rating
sns.boxplot(x='Rate', y='Price', data=df)
plt.title("Price Distribution by Rating")
plt.xlabel("Rating")
plt.ylabel("Price")
plt.show()
```

```python
# Count reviews per product
top_products = df['Product_name'].value_counts().head(10)

# Plot top 10 products
sns.barplot(x=top_products.values, y=top_products.index, palette="mako")
plt.title("Top 10 Products by Review Count")
plt.xlabel("Number of Reviews")
plt.ylabel("Product Name")
plt.show()
```

```python
# Filter products with poor ratings (1 or 2)
poor_rated = df[df['Rate'].isin([1, 2])]

# Count occurrences of each product
top_poor_products = poor_rated['Product_name'].value_counts().head(10)

# Plot top 10 poorly rated products
sns.barplot(x=top_poor_products.values, y=top_poor_products.index, palette="Reds")
plt.title("Top 10 Poorly Rated Products")
plt.xlabel("Number of Poor Ratings")
plt.ylabel("Product Name")
plt.show()
```
